# Diseño
1. Estrategia general de solución
El objetivo de la tienda es organizar una cantidad variable de precios para facilitar y agilizar las decisiones de venta. Para ello, primero se recibe la mercancía y se registra cada precio en una lista dinámica. Posteriormente, los precios se ordenan de acuerdo con la necesidad del día: de forma ascendente, para mostrar primero los productos de menor costo y atraer a los clientes, o de forma descendente, para identificar los productos más costosos y analizar mejor el margen del inventario.

2. Tipo de algoritmo
Para resolver este problema se utiliza un algoritmo de comparación directa, específicamente Insertion Sort (Ordenamiento por Inserción). Este método no necesita dividir el problema en partes complejas, como ocurre en divide y vencerás, ni utilizar tablas de almacenamiento propias de la programación dinámica.

Su funcionamiento puede compararse con la organización de una baraja de cartas: se toma un precio a la vez, se compara con los elementos que ya están ordenados y se coloca en la posición que le corresponde.

3. Tipo de implementación
Iterativa vs. Recursiva: La implementación es iterativa, ya que utiliza estructuras repetitivas como for y while para comparar y organizar los precios uno por uno. No es necesario que la función se llame a sí misma, como sucede en una implementación recursiva.

Incremental: Sí aplica. El algoritmo trabaja de manera incremental porque construye la solución progresivamente. En cada paso, mantiene una parte de la lista completamente ordenada y agrega un nuevo elemento en la posición que le corresponde dentro de esa sección.

4. Invariante del algoritmo (El principio que se mantiene)
El principio fundamental que garantiza el correcto funcionamiento del algoritmo es el siguiente: al comienzo de cada iteración del ciclo principal, todos los precios ubicados desde el inicio de la lista hasta la posición actual se encuentran ordenados correctamente.

De esta manera, el algoritmo únicamente debe tomar el nuevo precio y ubicarlo en la posición adecuada dentro de la parte que ya está ordenada. Este proceso se repite hasta que todos los precios del inventario quedan organizados según el criterio seleccionado.

```python

def ascendente(n, precios):
    for i in range(1, len(precios)):
        temp = precios[i]
        j = i - 1

        while j >= 0 and precios[j] > temp:
            precios[j+1] = precios[j]
            j -= 1

        precios[j+1] = temp

    return precios


def descendente(n, precios):
    for i in range(1, len(precios)):
        temp = precios[i]
        j = i - 1

        while j >= 0 and precios[j] < temp:
            precios[j+1] = precios[j]
            j -= 1

        precios[j+1] = temp

    return precios


def main():
    n = int(input("¿Cuántos precios desea insertar?: "))

    if n <= 0:
        print("Error: la cantidad de precios debe ser mayor que 0.")
        return

    precios = []

    for i in range(n):
        precio = int(input("Inserte Precio " + str(i+1) + ": "))

        if precio < 0:
            print("Error: el precio no puede ser negativo.")
            return

        precios.append(precio)

    ord = input(
        "¿Qué tipo de ordenamiento desea hacer? "
        "asc = Ascendente / des = Descendente: "
    )

    if ord == "asc":
        print(ascendente(n, precios))

    elif ord == "des":
        print(descendente(n, precios))

    else:
        print("Error: inserte un ordenamiento válido.")


main()
```
