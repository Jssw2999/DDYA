## 1) Descripción del problema **[1 pts]**


**Descripción**

La tienda necesita un pequeño sistema para organizar la atención al público. Existen tres necesidades independientes que, combinadas, describen el funcionamiento completo del negocio:

1. **Atención en orden de llegada:** los clientes deben ser atendidos exactamente en el orden en que llegaron (primero en llegar, primero en ser atendido). Esto corresponde naturalmente a una **cola (FIFO)**.
2. **Historial de atenciones y posibilidad de corregir errores:** cada vez que se atiende a alguien se debe poder "deshacer" esa atención (por ejemplo, si el empleado se equivocó de cliente). Deshacer siempre afecta a la *última* atención realizada, lo que corresponde a una **pila (LIFO)**.
3. **Catálogo de productos navegable en ambos sentidos:** el catálogo debe poder recorrerse de principio a fin y de fin a principio, además de permitir insertar y eliminar productos por código en cualquier posición. Esto corresponde a una **lista doblemente enlazada**.

El sistema integra estas tres estructuras en un único programa que un empleado de la tienda puede operar mediante un menú de opciones.

## 7) Análisis de complejidad **[5 pts]**
Incluye:
- Justificación
- Función T[n]
- Mejor y peor caso con gráfica


**Justificación**

Cada estructura fue implementada con enlaces (nodos), no con arreglos, por lo que no hay desplazamientos de elementos:

- **Cola de clientes** (lista enlazada simple con punteros a frente y final): encolar y atender son O(1) porque siempre se opera sobre los extremos. mostrar recorre todos los elementos: O(n).
- **Pila de atenciones** (lista enlazada simple con puntero a tope): apilar, desapilar y ver_tope son O(1). mostrar recorre toda la pila: O(n).
- **Lista doblemente enlazada de productos** (con punteros a primero y ultimo): agregar_final es O(1) gracias al puntero ultimo. mostrar_adelante y mostrar_atras son O(n). buscar y eliminar requieren recorrer la lista desde el inicio hasta encontrar el código, por lo que son O(n) en el caso general (una vez encontrado el nodo, desconectarlo es O(1)).

**Función T(n)**

Para las operaciones dominantes de búsqueda/eliminación/recorrido sobre una lista de *n* productos:

T(n) = c₁ + c₂·n

donde c₁ es el costo fijo (comparaciones, inicialización) y c₂ es el costo de avanzar un nodo. Esto se clasifica como **O(n)**.

Para las operaciones sobre los extremos de la cola y la pila (encolar, tender, apilar, desapilar):

T(n) = c₃ (constante, no depende de n) → **O(1)**

**Mejor y peor caso (búsqueda/eliminación por código)**

- **Mejor caso — O(1):** el producto buscado es el primero de la lista (primero); se encuentra en la primera comparación.
- **Peor caso — O(n):** el producto buscado es el último, o no existe en el catálogo; hay que recorrer los *n* nodos completos.

La gráfica de la sección de *Pruebas de desempeño* mide empíricamente el tiempo de buscar_producto para catálogos de tamaño creciente, confirmando el crecimiento lineal esperado.

```python

class NodoCliente:
    def __init__(self, nombre):
        self.nombre = nombre
        self.siguiente = None


class ColaClientes:
    def __init__(self):
        self.frente = None
        self.final = None

    def esta_vacia(self):
        return self.frente is None

    def encolar(self, nombre):
        nodo = NodoCliente(nombre)
        if self.esta_vacia():
            self.frente = self.final = nodo
        else:
            self.final.siguiente = nodo
            self.final = nodo

    def encolar_al_frente(self, nombre):
        """Usado cuando se deshace una atencion: el cliente vuelve a esperar primero."""
        nodo = NodoCliente(nombre)
        nodo.siguiente = self.frente
        self.frente = nodo
        if self.final is None:
            self.final = nodo

    def atender(self):
        if self.esta_vacia():
            return None
        nodo = self.frente
        self.frente = nodo.siguiente
        if self.frente is None:
            self.final = None
        return nodo.nombre

    def mostrar(self):
        nombres = []
        actual = self.frente
        while actual is not None:
            nombres.append(actual.nombre)
            actual = actual.siguiente
        return nombres



class NodoPila:
    def __init__(self, nombre):
        self.nombre = nombre
        self.siguiente = None


class PilaAtenciones:
    def __init__(self):
        self.tope = None

    def esta_vacia(self):
        return self.tope is None

    def apilar(self, nombre):
        nodo = NodoPila(nombre)
        nodo.siguiente = self.tope
        self.tope = nodo

    def desapilar(self):
        if self.esta_vacia():
            return None
        nodo = self.tope
        self.tope = nodo.siguiente
        return nodo.nombre

    def ver_tope(self):
        return self.tope.nombre if self.tope else None

    def mostrar(self):
        nombres = []
        actual = self.tope
        while actual is not None:
            nombres.append(actual.nombre)
            actual = actual.siguiente
        return nombres



class NodoProducto:
    def __init__(self, codigo, nombre, precio):
        self.codigo = codigo
        self.nombre = nombre
        self.precio = precio
        self.anterior = None
        self.siguiente = None


class ListaProductos:
    def __init__(self):
        self.primero = None
        self.ultimo = None

    def agregar_final(self, codigo, nombre, precio):
        nodo = NodoProducto(codigo, nombre, precio)
        if self.ultimo is None:
            self.primero = self.ultimo = nodo
        else:
            nodo.anterior = self.ultimo
            self.ultimo.siguiente = nodo
            self.ultimo = nodo

    def mostrar_adelante(self):
        productos = []
        actual = self.primero
        while actual is not None:
            productos.append((actual.codigo, actual.nombre, actual.precio))
            actual = actual.siguiente
        return productos

    def mostrar_atras(self):
        productos = []
        actual = self.ultimo
        while actual is not None:
            productos.append((actual.codigo, actual.nombre, actual.precio))
            actual = actual.anterior
        return productos

    def buscar(self, codigo):
        actual = self.primero
        while actual is not None:
            if actual.codigo == codigo:
                return (actual.codigo, actual.nombre, actual.precio)
            actual = actual.siguiente
        return None

    def _buscar_nodo(self, codigo):
        actual = self.primero
        while actual is not None:
            if actual.codigo == codigo:
                return actual
            actual = actual.siguiente
        return None

    def eliminar(self, codigo):
        nodo = self._buscar_nodo(codigo)
        if nodo is None:
            return False
        if nodo.anterior is not None:
            nodo.anterior.siguiente = nodo.siguiente
        else:
            self.primero = nodo.siguiente
        if nodo.siguiente is not None:
            nodo.siguiente.anterior = nodo.anterior
        else:
            self.ultimo = nodo.anterior
        return True



class SistemaTienda:
    def __init__(self):
        self.cola = ColaClientes()
        self.pila = PilaAtenciones()
        self.catalogo = ListaProductos()
    def registrar_clientes_iniciales(self, nombres):
        for nombre in nombres:
            self.cola.encolar(nombre)

    def agregar_cliente(self, nombre):
        self.cola.encolar(nombre)

    def atender_cliente(self):
        nombre = self.cola.atender()
        if nombre is not None:
            self.pila.apilar(nombre)
        return nombre

    def deshacer_atencion(self):
        nombre = self.pila.desapilar()
        if nombre is not None:
            self.cola.encolar_al_frente(nombre)
        return nombre

    def mostrar_espera(self):
        nombres = self.cola.mostrar()
        return " -> ".join(nombres) if nombres else "(sin clientes en espera)"

    def mostrar_ultimo_atendido(self):
        return self.pila.ver_tope()

    def mostrar_historial(self):
        nombres = self.pila.mostrar()
        return " -> ".join(nombres) if nombres else "(sin atenciones registradas)"


    def agregar_producto(self, codigo, nombre, precio):
        self.catalogo.agregar_final(codigo, nombre, precio)

    def mostrar_catalogo_adelante(self):
        return self.catalogo.mostrar_adelante()

    def mostrar_catalogo_atras(self):
        return self.catalogo.mostrar_atras()

    def buscar_producto(self, codigo):
        return self.catalogo.buscar(codigo)

    def eliminar_producto(self, codigo):
        return self.catalogo.eliminar(codigo)
```

```python

def menu():
    sistema = SistemaTienda()

    n = int(input("¿Cuántos clientes desea registrar?: "))
    nombres_iniciales = []
    for i in range(1, n + 1):
        nombres_iniciales.append(input(f"Ingrese el nombre del cliente {i}: "))
    sistema.registrar_clientes_iniciales(nombres_iniciales)

    opciones = """
MENU
1. Agregar cliente a la cola
2. Atender siguiente cliente
3. Mostrar clientes en espera
4. Mostrar ultimo cliente atendido
5. Deshacer ultima atencion
6. Agregar producto al catalogo
7. Mostrar catalogo (primero a ultimo)
8. Mostrar catalogo (ultimo a primero)
9. Buscar producto por codigo
10. Eliminar producto por codigo
0. Salir
"""

    while True:
        print(opciones)
        opcion = input("Seleccione una opcion: ")

        if opcion == "1":
            nombre = input("Ingrese el nuevo cliente: ")
            sistema.agregar_cliente(nombre)
            print("Cola actualizada:", sistema.mostrar_espera())

        elif opcion == "2":
            nombre = sistema.atender_cliente()
            if nombre is None:
                print("No hay clientes en espera.")
            else:
                print(f"Atendiendo a: {nombre}")
                print("Clientes esperando:", sistema.mostrar_espera())

        elif opcion == "3":
            print("Clientes esperando:", sistema.mostrar_espera())

        elif opcion == "4":
            ultimo = sistema.mostrar_ultimo_atendido()
            print("Ultimo atendido:", ultimo if ultimo else "(sin atenciones)")

        elif opcion == "5":
            nombre = sistema.deshacer_atencion()
            if nombre is None:
                print("No hay atenciones que deshacer.")
            else:
                print(f"Se deshizo la atencion de: {nombre}")
                print("Clientes esperando:", sistema.mostrar_espera())

        elif opcion == "6":
            codigo = int(input("Codigo: "))
            nombre = input("Nombre: ")
            precio = float(input("Precio: "))
            sistema.agregar_producto(codigo, nombre, precio)
            print("Producto agregado.")

        elif opcion == "7":
            for c, n_, p in sistema.mostrar_catalogo_adelante():
                print(f"{c} - {n_} - ${p}")

        elif opcion == "8":
            for c, n_, p in sistema.mostrar_catalogo_atras():
                print(f"{c} - {n_} - ${p}")

        elif opcion == "9":
            codigo = int(input("Codigo a buscar: "))
            resultado = sistema.buscar_producto(codigo)
            print("Encontrado:", resultado if resultado else "no existe")

        elif opcion == "10":
            codigo = int(input("Codigo a eliminar: "))
            ok = sistema.eliminar_producto(codigo)
            print("Eliminado correctamente." if ok else "Codigo no encontrado.")

        elif opcion == "0":
            print("Fin del programa.")
            break
        else:
            print("Opcion invalida.")

menu()
```

#### Código base para benchmarking

```python


def bench(n: int, step: int = 200) -> list[tuple[int, float]]:
    print("Generando datos ...", end='')
    import time
    results: list[tuple[int, float]] = []

    for size in range(step, n, step):
        catalogo_prueba = SistemaTienda()
        for codigo in range(size):
            catalogo_prueba.agregar_producto(codigo, f"Producto{codigo}", 1000 + codigo)

        codigo_a_buscar = size - 1  # peor caso: el ultimo elemento de la lista

        ti = time.perf_counter()
        catalogo_prueba.buscar_producto(codigo_a_buscar)
        tf = time.perf_counter()

        delta = (tf - ti) * 1000
        print(".", end="")
        results.append((size, delta))
    return results


def dibujar(x, y, dimensiones=(7, 5)):
    assert len(x) == len(y)
    import matplotlib.pyplot as plt
    plt.figure(figsize=dimensiones)
    plt.plot(x, y, marker='o')
    plt.xlabel("Tamaño del catalogo (n)")
    plt.ylabel("Tiempo de busqueda (ms)")
    plt.title("Tiempo de buscar_producto en peor caso vs. tamaño del catalogo")
    plt.show()


tiempos = bench(4000, step=200)

x = [size for size, timing in tiempos]
y = [timing for size, timing in tiempos]

dibujar(x, y)

print("\nComo se observa, el tiempo crece de forma aproximadamente lineal con n, "
      "lo cual coincide con el analisis teorico O(n) para la busqueda en la lista "
      "doblemente enlazada.")
```
