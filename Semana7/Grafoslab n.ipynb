### 5) Análisis de complejidad **[15 pts]**
La complejidad temporal de los algoritmos BFS y DFS es O(n + e).
El componente n (vértices) representa el tiempo para visitar cada centro de distribución una sola vez.
El componente e (aristas) corresponde al tiempo necesario para recorrer todas las conexiones entre ellos.
Por lo tanto, el tiempo de ejecución final depende tanto de la cantidad de nodos como de la densidad de sus rutas.

```python
# Escribe tu código aquí.
from collections import deque
 
 
class RedDistribucion:
 
    def __init__(self):
        self.grafo = {
            "A": ["B", "C"],
            "B": ["A", "D", "E"],
            "C": ["A", "F"],
            "D": ["B"],
            "E": ["B", "G"],
            "F": ["C", "G"],
            "G": ["E", "F"],
        }
 
    
    def mostrar_conexiones(self):
        print("\n--- Lista de Adyacencia ---")
        for centro, vecinos in self.grafo.items():
            print(f"{centro}: {', '.join(vecinos)}")
 
    def ejecutar_bfs(self, centro_inicial):
        visitados = set()
        cola = deque()
 
        # Agregar centro inicial a la cola
        cola.append((centro_inicial, 0))  # (nodo, nivel)
        visitados.add(centro_inicial)
 
        recorrido = []
        niveles = {}
 
        # Bucle BFS (Extraer centro de la cola -> Visitar vecinos no visitados -> Agregar vecinos a la cola)
        while cola:
            centro_actual, nivel = cola.popleft()
            recorrido.append(centro_actual)
 
            if nivel not in niveles:
                niveles[nivel] = []
            niveles[nivel].append(centro_actual)
 
            # Visitar vecinos no visitados y agregarlos a la cola
            for vecino in self.grafo[centro_actual]:
                if vecino not in visitados:
                    visitados.add(vecino)
                    cola.append((vecino, nivel + 1))
 
        print(f"\nRecorrido BFS desde {centro_inicial}:")
        print(" -> ".join(recorrido))
        print("Niveles:")
        for n, nodos in niveles.items():
            print(f"  Nivel {n}: {', '.join(nodos)}")
 
    def ejecutar_dfs(self, centro_inicial):
        visitados = set()
        recorrido = []
 
        
        def dfs_recursivo(centro_actual):
            visitados.add(centro_actual)
            recorrido.append(centro_actual)
 
            for vecino in self.grafo[centro_actual]:
                if vecino not in visitados:
                    dfs_recursivo(vecino)
 
        dfs_recursivo(centro_inicial)
 
        # Mostrar recorrido DFS
        print(f"\nRecorrido DFS desde {centro_inicial}:")
        print(" -> ".join(recorrido))
 
    def buscar_centro(self):
        
        inicio = (
            input("Ingrese el centro inicial para la búsqueda: ")
            .strip()
            .upper()
        )
        if inicio not in self.grafo:
            print("Mostrar mensaje de error: El centro inicial no existe.")
            return
 
        objetivo = (
            input("Ingresar centro objetivo que desea localizar: ")
            .strip()
            .upper()
        )
        if objetivo not in self.grafo:
            print("Mostrar mensaje de error: El centro objetivo no existe.")
            return
 
        # Ejecutar BFS
        visitados = set([inicio])
        cola = deque([inicio])
        padre = {inicio: None}
        encontrado = False
 
        while cola:
            actual = cola.popleft()
 
            # ¿Centro encontrado?
            if actual == objetivo:
                encontrado = True
                break
 
            for vecino in self.grafo[actual]:
                if vecino not in visitados:
                    visitados.add(vecino)
                    padre[vecino] = actual
                    cola.append(vecino)
 
        # Decisión: ¿Centro encontrado?
        if encontrado:
            # Reconstruir ruta
            ruta = []
            curr = objetivo
            while curr is not None:
                ruta.append(curr)
                curr = padre[curr]
            ruta.reverse()
 
            print(f"\nCentro encontrado.")
            print(f"Ruta: {' -> '.join(ruta)}")
            print(f"Cantidad de conexiones: {len(ruta) - 1}")
        else:
            print("\nMostrar: No existe una ruta hasta el centro solicitado.")
 
    def agregar_centro(self):
        nuevo = input("\nIngresar nombre del nuevo centro: ").strip().upper()
 
        #¿El centro ya existe?
        if nuevo in self.grafo:
            print(
                "Mostrar mensaje de error: El centro ya existe en el grafo."
            )
            return
 
        self.grafo[nuevo] = []
 
        # Ingresar centros con los que se conecta
        print(
            f"Ingrese los centros a los que se conecta '{nuevo}' (separados por coma):"
        )
        conexiones = input("Centros: ").strip().upper().split(",")
 
        for c in conexiones:
            c = c.strip()
            if c in self.grafo and c != nuevo:
                self.grafo[nuevo].append(c)
                self.grafo[c].append(nuevo)
 
        # Mostrar lista de adyacencia
        self.mostrar_conexiones()
 
        print(f"\nEjecutando BFS y DFS desde el nuevo centro ({nuevo}):")
        self.ejecutar_bfs(nuevo)
        self.ejecutar_dfs(nuevo)
 
 
# Flujo Principal del Diagrama
def programa_principal():
    
    red = RedDistribucion()
 
    red.mostrar_conexiones()
 
    while True:
        print("\n" + "=" * 40)
        print("¿Qué desea hacer el usuario?")
        print("1. BFS / DFS")
        print("2. Buscar centro")
        print("3. Agregar centro")
        print("4. Salir (FIN)")
 
        opcion = input("Seleccione una opción: ").strip()
 
        if opcion == "1":
            # Seleccionar centro inicial
            centro = (
                input("\nSeleccionar centro inicial: ").strip().upper()
            )
 
            #El centro existe?
            if centro in red.grafo:
                # Ejecutar BFS
                red.ejecutar_bfs(centro)
                # Ejecutar DPS
                red.ejecutar_dfs(centro)
            else:
                #retorna al menú
                print(
                    "\nMostrar mensaje de error: El centro ingresado no existe."
                )
 
        elif opcion == "2":
            red.buscar_centro()
 
        elif opcion == "3":
            red.agregar_centro()
 
        elif opcion == "4":
            # FIN
            print("\nFIN del programa.")
            break
        else:
            print("Opción no válida. Intente nuevamente.")
 
 
if __name__ == "__main__":
    programa_principal()
```
