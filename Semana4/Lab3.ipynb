## 1) Descripción del problema **[1 pts]**

Una tienda en Colombia necesita desarrollar un programa que calcule la menor cantidad de monedas necesarias para entregar un cambio determinado a un cliente.

monedas = [50, 100, 200, 500, 1000]

Ingrese el valor del cambio: 1850

## 7) Análisis de complejidad **[5 pts]**

### Justificación

La función `calc_min(val_camb, memo)` utiliza **memorización**. Cada valor de cambio que se calcula se guarda en el diccionario `memo`, por lo que no es necesario volver a resolverlo.

Hay 5 denominaciones posibles: `50, 100, 200, 500 y 1000`. Para cada valor de cambio se recorren como máximo esas 5 monedas. Como el número de denominaciones es constante, el costo por estado es constante.

### Función de tiempo

Sea `n` el valor del cambio. Los valores que pueden aparecer son aproximadamente:

`0, 50, 100, ..., n`

Por tanto, existen aproximadamente `n/50` estados y en cada estado se revisan 5 monedas:

**T(n) = 5(n/50) + c**

Como 5 y 50 son constantes:

**T(n) = O(n)**

### Mejor caso

Si `val_camb = 0`, la función retorna inmediatamente:

**T(n) = O(1)**

También puede ocurrir que un valor ya esté almacenado en `memo`, en cuyo caso la consulta es de tiempo constante promedio.

### Peor caso

Para un valor nuevo de cambio, se calculan todos los estados necesarios hasta llegar a 0 y se revisan las 5 monedas en cada estado:

**T(n) = O(n)**

### Complejidad espacial

El diccionario `memo` puede almacenar aproximadamente `n/50` resultados, por lo que el espacio utilizado es:

**S(n) = O(n)**

La recursión también utiliza una pila proporcional al número de estados: **O(n)**.

El mejor caso es prácticamente constante, mientras que el peor caso crece linealmente con el valor del cambio.


## 8) Código funcional **[10 pts]**

El siguiente código utiliza **recursividad y memorización**. El diccionario `memo` almacena los resultados ya calculados para evitar repetir operaciones.


````python
def calc_min(val_camb, memo):
    monedas = [50, 100, 200, 500, 1000]

    # Caso base
    if val_camb == 0:
        return 0

    # Si ya calculamos este valor, reutilizamos el resultado
    if val_camb in memo:
        return memo[val_camb]

    minimo = float("inf")

    # Probar cada denominación disponible
    for moneda in monedas:
        if moneda <= val_camb:
            resp = calc_min(val_camb - moneda, memo) + 1

            if resp < minimo:
                minimo = resp

    # Guardar el resultado para no recalcularlo
    memo[val_camb] = minimo

    return minimo


def main():
    memo = {}

    while True:
        try:
            val_camb = int(input("Ingrese el valor del cambio: "))

            if val_camb < 0:
                print("Ingrese un valor positivo.")
                continue

            break

        except ValueError:
            print("Ingrese solo números enteros.")

    if val_camb % 50 != 0:
        print("No se puede entregar el cambio exactamente con las monedas disponibles.")
    else:
        resp = calc_min(val_camb, memo)
        print("Cantidad mínima de monedas:", resp)


main()
````
