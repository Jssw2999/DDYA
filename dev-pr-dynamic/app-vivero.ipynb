```python
# Escribe tu código aquí.
memo = {}

def vivero(n):
    if n == 0:
        return n
    if n == 1:
        return n
    
    if n not in memo:
        memo[n] = vivero(n-1) + vivero(n-2)

    return memo[n]
    
    

def main():
    proceed = False
    while proceed == False:
         n = int(input("Inserte el mes en el que desea consultar el numero de plantas nuevas: "))
        
         if n > -1:
             proceed = True
         else:
            print("Introduzca un numero entero positivo")
        
    print(vivero(n))

main()
```
