## 8) Código funcional **[15 pts]**

```python
def merge_sort(codes):
    if len(codes) <= 1:
        return codes

    med = len(codes) // 2
    izq = codes[:med]
    der = codes[med:]

    izq_ord = merge_sort(izq)
    der_ord = merge_sort(der)

    return merge(izq_ord, der_ord)


def merge(izq_ord, der_ord):
    resp = []
    i = 0
    j = 0

    while i < len(izq_ord) and j < len(der_ord):
        if izq_ord[i] < der_ord[j]:
            resp.append(izq_ord[i])
            i += 1
        else:
            resp.append(der_ord[j])
            j += 1

    while i < len(izq_ord):
        resp.append(izq_ord[i])
        i += 1

    while j < len(der_ord):
        resp.append(der_ord[j])
        j += 1

    return resp


def bus_bin(codes, obj):
    ini = 0
    fin = len(codes) - 1

    while ini <= fin:
        med = (ini + fin) // 2

        if codes[med] == obj:
            return med
        elif codes[med] < obj:
            ini = med + 1
        else:
            fin = med - 1

    return -1


def main():
    n = int(input("Ingrese la cantidad de productos: "))

    if n <= 0:
        print("La cantidad de productos debe ser mayor que 0.")
        return

    codes = []

    for i in range(n):
        temp = int(input("Introduzca el codigo del producto " + str(i + 1) + ": "))

        if temp < 0:
            print("El codigo del producto no puede ser negativo.")
            return

        codes.append(temp)

    codes_ord = merge_sort(codes)

    obj = int(input("Ingrese el codigo que desea buscar: "))

    if obj < 0:
        print("El codigo a buscar no puede ser negativo.")
        return

    busq = bus_bin(codes_ord, obj)

    if busq != -1:
        print("El producto con el codigo " + str(obj) + " fue encontrado.")
    else:
        print("El producto con el codigo " + str(obj) +
              " no se encuentra registrado.")


main()
```
